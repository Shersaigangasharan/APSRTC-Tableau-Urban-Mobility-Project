```python
import pandas as pd

# 1. Loading the dataset
df = pd.read_csv("/content/drive/MyDrive/APSRTC/APSRTC_Transport_Data.csv")
```

```python
# 2. Viewing the basic structure
print("Shape:", df.shape)
print(df.head())
print(df.info())
```

```python
# 3. Checking the missing values
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# 5. Converting the date column to datetime
df['date'] = pd.to_datetime(df['date'], errors='coerce')
```

```python
# 6. Converting the numeric columns
numeric_cols = [
    'capacity', 'passengers', 'occupancy_rate', 'distance_km',
    'fare_per_passenger', 'revenue', 'fuel_consumed_liters'
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
# 7. Creating the new calculated fields
df['profit'] = df['revenue'] - (df['fuel_consumed_liters'] * 110)  # approx diesel cost
df['fuel_efficiency_km_per_liter'] = df['distance_km'] / df['fuel_consumed_liters']
df['revenue_per_km'] = df['revenue'] / df['distance_km']
```

```python
# 8. Checking for the remaining issues
print("\nRemaining Missing Values:")
print(df.isnull().sum())
```

```python
# 9. Saving the cleaned dataset
df.to_csv("/content/drive/MyDrive/APSRTC/APSRTC_cleaned.csv", index=False)

print("\nCleaning complete. File saved to Google Drive as /content/drive/MyDrive/APSRTC/APSRTC_cleaned.csv")
```
